```python
# Cell 1: Complete Workflow
"""
Complete Workflow Example
=========================
Demonstrates full data processing pipeline.
"""

import pandas as pd
from chemdata.validation import ChemicalValidator, NumericalValidator
from chemdata.normalization import UnitNormalizer, ChemicalNormalizer
from chemdata.quality import OutlierDetector, MissingDataAnalyzer
from chemdata.provenance import ProvenanceTracker
from chemdata.export import DataExporter

# Cell 2: Load sample data
data = {
    'catalyst': ['TiO2', 'Ti O2', 'ZnO', 'Fe2O3'],
    'concentration_mg_L': [10.0, 20.0, 15.0, 25.0],
    'pH': [7.0, 6.5, 8.0, 7.5],
    'temperature_C': [25.0, 30.0, 25.0, 35.0],
    'degradation_percent': [85.0, 92.0, 78.0, 88.0],
}
df = pd.DataFrame(data)

# Cell 3: Add provenance
tracker = ProvenanceTracker()
df = tracker.add_provenance(
    df,
    source="Example Dataset",
    source_type="synthetic",
    extraction_date="2024-01-01"
)

# Cell 4: Normalize chemistry
chem_normalizer = ChemicalNormalizer()
df = chem_normalizer.normalize_dataframe_column(
    df,
    'catalyst',
    normalization_type='formula'
)

print("Normalized catalysts:")
print(df[['catalyst', 'catalyst_normalized']])

# Cell 5: Normalize units
unit_normalizer = UnitNormalizer()
df = unit_normalizer.normalize_column(
    df,
    'temperature_C',
    from_unit='C',
    to_unit='K',
    conversion_type='temperature'
)

print("\nTemperature conversion:")
print(df[['temperature_C', 'temperature_C_K']])

# Cell 6: Detect outliers
outlier_detector = OutlierDetector()
df = outlier_detector.mark_outliers(
    df,
    'degradation_percent',
    methods=['zscore', 'iqr']
)

print("\nOutlier detection:")
print(df[['degradation_percent', 'degradation_percent_outlier_any']])

# Cell 7: Missing data analysis
missing_analyzer = MissingDataAnalyzer()
missing_report = missing_analyzer.generate_missing_report(df)

print("\nMissing data report:")
print(f"Total cells: {missing_report['total_cells']}")
print(f"Missing cells: {missing_report['missing_cells']}")

# Cell 8: Export clean data
exporter = DataExporter()
exporter.export_csv(df, 'examples/clean_data.csv', include_provenance=True)

print("\nData exported to: examples/clean_data.csv")
```
